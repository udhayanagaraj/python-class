```python
from threading import Thread

class Example(Thread):
    def run(self):
        for i in range(10):
            print("i =", i)

e1 = Example()
e2 = Example()

e1.start()
e2.start()

e1.join()
e2.join()
```

```python
import asyncio

async def cor1():
    print("cor1 start")
    for i in range(10):
        await asyncio.sleep(1.5)
        print("cor1", i)

async def cor2():
    print("cor2 start")
    for i in range(15):
        await asyncio.sleep(1)
        print("cor2", i)

async def main():
    await asyncio.gather(cor1(), cor2())

asyncio.run(main())
```

```python
def giveHint(number, guess):
    if guess == number:
        return "Right"

    diff = abs(number - guess)

    if diff > 10:
        return "Cold"
    elif diff <= 10:
        return "Hot"
    
giveHint(50,39)
```

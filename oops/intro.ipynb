```python
class Student:
    def __init__(self,first_name,last_name,age):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age

    def print_name(self):
        print(self.first_name)

    def update_name(self,first_name):
        self.first_name = first_name


student = Student("udhaya","nagaraj",25)
student.print_name()
student.update_name("Udhaya N")
student.print_name()
```

```python
data = []

for i in range(10):
    data.append(Student(f"{i}",f"{i}*{i}",1*100))
```

```python
for d in data:
    d.print_name()
```

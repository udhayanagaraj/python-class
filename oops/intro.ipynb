```python
class School:
    def __init__(self,name,address):
        self.name = name
        self.address = address
        self.headmaster = None
    
    def set_headmaster(self,staff):
        self.headmaster = staff

    def print_headmaster(self):
        return self.headmaster
    
    def get_headmaster_age(self):
        if self.headmaster:
            print("Age of master: ",self.headmaster.print_age())
        else:
            print("Please add a headmaster first.")


class Staff:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    def print_age(self):
        return self.age

    def __str__(self):
        return f"Name: {self.name}, Age: {self.age}"
    

sch = School("PSG","Chennai")
staff =  Staff("Udhaya",24)
sch.set_headmaster(staff)
print(sch.print_headmaster())
sch.get_headmaster_age()
```

```python
from abc import ABC, abstractmethod


class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Bark bark")

class Cat(Animal):
    def sound(self):
        print("Meow")

d = Dog()
d.sound()

c = Cat()
c.sound()
```

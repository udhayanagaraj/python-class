```python


nums=[34,56,67,67,763,4563]

def find_largest_three(lst):
    m1 = m2 = m3 = 0

    for i in lst:
        if i > m1:
            m3 = m2
            m2 = m1
            m1 = i
        elif i < m1 and i != m2:
            m3 = m2
            m2 = i
        elif i < m2 and i != m3:
            m3 = i

    return [m1,m2,m3]

print(find_largest_three(nums))

```

```python
def reverse_number(n):
    temp = n
    ans = 0
    
    while temp != 0:
        r = temp % 10
        ans = ans * 10 + r
        temp = temp // 10
    return ans

print(reverse_number(123450))
```
